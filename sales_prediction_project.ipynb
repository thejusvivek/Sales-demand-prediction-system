```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score,accuracy_score,classification_report
```

```python
df= pd.read_csv('sales_data_sample.csv',encoding='latin1')
```

```python
# 1. Select relevant columns and drop missing rows
model_df = df[['QUANTITYORDERED', 'PRICEEACH', 'SALES']].dropna()
X = model_df[['QUANTITYORDERED', 'PRICEEACH']] # Features

# ==========================================
# 2. LINEAR REGRESSION (Predicting Sales Value)
# ==========================================
y_reg = model_df['SALES']
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X, y_reg, test_size=0.2, random_state=42)

lin_reg = LinearRegression()
lin_reg.fit(X_train_r, y_train_r)
y_pred_r = lin_reg.predict(X_test_r)

print("--- Linear Regression Results ---")
print(f"Mean Squared Error (MSE): {mean_squared_error(y_test_r, y_pred_r):.2f}")
print(f"R2 Score: {r2_score(y_test_r, y_pred_r):.4f}\n")

# ==========================================
# 3. LOGISTIC REGRESSION (Demand Classification)
# ==========================================
median_sales = model_df['SALES'].median()
y_clf = (model_df['SALES'] > median_sales).astype(int)

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X, y_clf, test_size=0.2, random_state=42)

log_reg = LogisticRegression()
log_reg.fit(X_train_c, y_train_c)
y_pred_c = log_reg.predict(X_test_c)

print("--- Logistic Regression Results ---")
print(f"Accuracy Score: {accuracy_score(y_test_c, y_pred_c) * 100:.2f}%")
print("Classification Report:")
print(classification_report(y_test_c, y_pred_c))
```

```python
import pickle

# 1. Save the trained Linear Regression model
with open('linear_regression_model.pkl', 'wb') as f:
    pickle.dump(lin_reg, f)

# 2. Save the trained Logistic Regression model
with open('logistic_regression_model.pkl', 'wb') as f:
    pickle.dump(log_reg, f)

print("Models saved successfully as .pkl files!")
```
